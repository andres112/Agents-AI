````python
# Set a list of models to use with clear purpose of their strengths
models = [
		"gpt-4o-mini",               # Strong reasoning and creativity
		"anthropic/claude-3-haiku",  # Concise and poetic expression
		"google/gemini-2.0-flash-lite-001",  # Broad knowledge and context understanding
		"deepseek/deepseek-chat-v3.1",       # Analytical and detailed responses
		"x-ai/grok-4-fast",          # Programming and technical expertise
		"meta-llama/llama-3.2-3b-instruct:free",  # Latest advancements in reasoning
		"google/gemma-3-12b-it" 			# Multimodal capabilities
]
# Create the initial request of the plan
request = f""" Come up with a idea for a project to solve a complex problem of logistics. 
Second with the steps or phases needed to implement the project.

Result should be in JSON format with each step clearly defined and explained. With following structure:
{{"idea": "project idea", "phases": ["phase_1", "phase_2", "phase_3", ...]}}
"""
request += "Answer only with the idea and phases, no explanation or elaboration."
messages = [{"role": "user", "content": request + ". Answer in 300 words or less"}]
````

````python
# Generate the plan that will be executed by the orchestrator
plan = openai.chat.completions.create(
		model="gpt-4.1-nano",
		messages=messages,
)
plan = plan.choices[0].message.content
print(plan)
````

````python
# Define the orchestrator that will execute the plan selecting the best model for each part
C = {
	"models": models,
	"selected_models": [],
	"integrated_outputs": []
}

plan_response = json.loads(plan)

orchestrator_prompt = f"""You are an orchestrator that will create a plan to implement a project for the idea: {plan_response['idea']}.
You have access to the following models, each with their own strengths: {', '.join(C['models'])}.
You will analyze the idea and select the best model for each phase of the plan.
Here are the phases:

{plan_response['phases']}

You will respond with the selected model and the output for each phase of the plan in JSON object with the following format:
{{"selected_models": ["model for phase 1", "model for phase 2", ...],
"integrated_outputs": ["output from model for phase 1", "output from model for phase 2", ...]}}

Not other format is allowed, remove all Markdown format in response.
"""
messages = [{"role": "user", "content": orchestrator_prompt}]
response = openRouter.chat.completions.create(
		model="openai/gpt-4o-mini",
		messages=messages,
)
C_response = response.choices[0].message.content

display(Markdown(f"```json\n{C_response}\n```"))
````

````python
# Set Orchestrator response
C["selected_models"] = json.loads(C_response)["selected_models"]
C["integrated_outputs"] = json.loads(C_response)["integrated_outputs"]
````

````python
for index, step in enumerate(plan_response['phases']):
	print(f"{step}")
````

````python
# Now we can execute the plan based on the orchestrator's selections
final_plan = []
for index, model in enumerate(C["selected_models"]):
		prompt = C["integrated_outputs"][index]
		messages = [{"role": "user", "content": prompt + ". Answer in 200 words or less"}]
		response = openRouter.chat.completions.create(
				model=model,
				messages=messages
		)
		output = response.choices[0].message.content
		final_plan.append({
			"step_or_phase": plan_response['phases'][index],
			"output": output,
		})

		rows = [f"| {plan_response['phases'][index]}  | {model} | {escape_md(output)} |"]
		table = "| Part | Model | Output |\n|-------|--------|--------|\n" + "\n".join(rows)
		display(Markdown(table))
````

````python
print(final_plan)
````

````python
# following the orchestrator pattern design, create a synthesizer to generate a unified output

parts_text = ""
for part in final_plan:
	parts_text += f"{part['step_or_phase']}\nContent: {part['output']}\n\n"

synth__prompt = f"""You are a synthesizer that will create a unified and fluid output from multiple parts.
You have the following parts to synthesize in order to create a final report and plan:

{parts_text}

The response should be a comprehensive report that integrates all parts into a coherent whole, not merely a summary of each part and assign estimated timelines for each phase.

The report should address the key objectives outlined in the prompt and provide actionable recommendations based on the synthesized information.

The answer in Markdown format with a clear structure and use headings and subheadings as needed.
Not more than 1000 words.
"""
messages = [{"role": "user", "content": synth__prompt}]
response = openai.chat.completions.create(
	model="gpt-5-nano",
	messages=messages,
)
synth_response = response.choices[0].message.content
display(Markdown(synth_response))
````
